```python

# ============================================================
# Crack404 AI/ML Internship - Week 3
# Project: Automated Invoice & Receipt Digitization
#
# PHASE 2 - CELL 1
# Image Loading and Grayscale Conversion
# ============================================================

from pathlib import Path

import cv2
import numpy as np
import matplotlib.pyplot as plt


# Step 1: Define project paths
PROJECT_DIR = Path(r"D:\Ai_Ml_intren\week_3")

RAW_DIR = PROJECT_DIR / "raw_images"

GRAYSCALE_DIR = (
    PROJECT_DIR / "outputs" / "phase2_grayscale"
)

GRAYSCALE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# Step 2: Collect receipt images
supported_formats = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

image_files = sorted([
    file for file in RAW_DIR.iterdir()
    if file.is_file()
    and file.suffix.lower() in supported_formats
])

print("=" * 60)
print("PHASE 2 - CELL 1: GRAYSCALE CONVERSION")
print("=" * 60)

print("\nTotal Images Found:", len(image_files))

assert 3 <= len(image_files) <= 5, (
    "Expected 3-5 receipt images."
)


# Step 3: Create comparison figure
fig, axes = plt.subplots(
    2,
    len(image_files),
    figsize=(17, 8)
)

processed_count = 0


# Step 4: Process each receipt image
for index, image_path in enumerate(image_files):

    # Load image (supports long Windows file names)
    image_bytes = np.fromfile(
        str(image_path),
        dtype=np.uint8
    )

    image = cv2.imdecode(
        image_bytes,
        cv2.IMREAD_COLOR
    )

    if image is None:
        raise ValueError(
            f"Cannot read image: {image_path.name}"
        )

    # Convert BGR image to grayscale
    gray_image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    # Convert BGR to RGB for visualization
    image_rgb = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    # Save grayscale image
    output_path = GRAYSCALE_DIR / (
        f"receipt_{index + 1:02d}_grayscale.png"
    )

    saved = cv2.imwrite(
        str(output_path),
        gray_image
    )

    if not saved:
        raise IOError(
            f"Could not save: {output_path}"
        )

    processed_count += 1

    # Display original image
    axes[0, index].imshow(image_rgb)
    axes[0, index].set_title(
        f"Receipt {index + 1} - Original"
    )
    axes[0, index].axis("off")

    # Display grayscale image
    axes[1, index].imshow(
        gray_image,
        cmap="gray",
        vmin=0,
        vmax=255
    )

    axes[1, index].set_title(
        f"Receipt {index + 1} - Grayscale"
    )
    axes[1, index].axis("off")

    # Print image information
    height, width = image.shape[:2]

    print(f"\nReceipt {index + 1}")
    print("Original File:", image_path.name)
    print("Resolution:", f"{width} x {height}")
    print("Grayscale Shape:", gray_image.shape)
    print("Saved As:", output_path.name)


# Step 5: Show comparison
plt.suptitle(
    "Week 3 - Original vs Grayscale Receipts",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()


# Step 6: Final verification
print("\n" + "=" * 60)

print("Images Processed:", processed_count)
print("Images Saved:", len(list(
    GRAYSCALE_DIR.glob("*.png")
)))

print("Output Directory:", GRAYSCALE_DIR)

assert processed_count == len(image_files)

print("\nPHASE 2 - CELL 1 COMPLETED SUCCESSFULLY!")
```

```python
# ============================================================
# Crack404 AI/ML Internship - Week 3
# Project: Automated Invoice & Receipt Digitization
#
# PHASE 2 - CELL 2
# Gaussian Blur / Noise Reduction
# ============================================================

from pathlib import Path
import cv2
import matplotlib.pyplot as plt

# Step 1: Define input and output directories
PROJECT_DIR = Path(r"D:\Ai_Ml_intren\week_3")

GRAYSCALE_DIR = PROJECT_DIR / "outputs" / "phase2_grayscale"
BLUR_DIR = PROJECT_DIR / "outputs" / "phase2_blur"

BLUR_DIR.mkdir(parents=True, exist_ok=True)

# Step 2: Collect grayscale images
gray_files = sorted(GRAYSCALE_DIR.glob("*.png"))

print("=" * 60)
print("PHASE 2 - CELL 2: GAUSSIAN BLUR")
print("=" * 60)

print("\nTotal Grayscale Images Found:", len(gray_files))

assert len(gray_files) >= 3, "At least 3 grayscale images required."

# Step 3: Create comparison plot
fig, axes = plt.subplots(2, len(gray_files), figsize=(17, 8))

processed_count = 0

# Step 4: Process each grayscale image
for index, image_path in enumerate(gray_files):

    gray_image = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)

    if gray_image is None:
        raise ValueError(f"Cannot read image: {image_path.name}")

    # Apply Gaussian Blur
    blurred_image = cv2.GaussianBlur(gray_image, (3, 3), 0)

    # Save blurred image
    output_path = BLUR_DIR / f"receipt_{index + 1:02d}_blur.png"

    saved = cv2.imwrite(str(output_path), blurred_image)

    if not saved:
        raise IOError(f"Could not save: {output_path}")

    processed_count += 1

    # Show grayscale image
    axes[0, index].imshow(gray_image, cmap="gray", vmin=0, vmax=255)
    axes[0, index].set_title(f"Receipt {index + 1} - Grayscale")
    axes[0, index].axis("off")

    # Show blurred image
    axes[1, index].imshow(blurred_image, cmap="gray", vmin=0, vmax=255)
    axes[1, index].set_title(f"Receipt {index + 1} - Blurred")
    axes[1, index].axis("off")

    print(f"\nReceipt {index + 1}")
    print("Input File :", image_path.name)
    print("Output File:", output_path.name)
    print("Shape      :", blurred_image.shape)

# Step 5: Display comparison
plt.suptitle(
    "Week 3 - Grayscale vs Gaussian Blur",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# Step 6: Final verification
print("\n" + "=" * 60)
print("Images Processed:", processed_count)
print("Images Saved:", len(list(BLUR_DIR.glob('*.png'))))
print("Output Directory:", BLUR_DIR)

assert processed_count == len(gray_files)

print("\nPHASE 2 - CELL 2 COMPLETED SUCCESSFULLY!")
```

```python
# ============================================================
# Crack404 AI/ML Internship - Week 3
# Project: Automated Invoice & Receipt Digitization
#
# PHASE 2 - CELL 3
# Global Thresholding and Adaptive Thresholding
# ============================================================

from pathlib import Path
import cv2
import matplotlib.pyplot as plt

# Step 1: Define directories
PROJECT_DIR = Path(r"D:\Ai_Ml_intren\week_3")

BLUR_DIR = PROJECT_DIR / "outputs" / "phase2_blur"
THRESHOLD_DIR = PROJECT_DIR / "outputs" / "phase2_threshold"

THRESHOLD_DIR.mkdir(parents=True, exist_ok=True)

# Step 2: Collect blurred images
blur_files = sorted(BLUR_DIR.glob("*.png"))

print("=" * 65)
print("PHASE 2 - CELL 3: THRESHOLDING")
print("=" * 65)

print("\nTotal Blurred Images Found:", len(blur_files))

assert len(blur_files) >= 3, "At least 3 blurred images required."

processed_count = 0

# Step 3: Process each image one by one
for index, image_path in enumerate(blur_files, start=1):

    blurred_image = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)

    if blurred_image is None:
        raise ValueError(f"Cannot read image: {image_path.name}")

    # Step 4: Apply Global Thresholding
    _, global_thresh = cv2.threshold(
        blurred_image,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Step 5: Apply Adaptive Thresholding
    adaptive_thresh = cv2.adaptiveThreshold(
        blurred_image,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11,
        2
    )

    # Step 6: Save thresholded images
    global_output_path = THRESHOLD_DIR / f"receipt_{index:02d}_global.png"
    adaptive_output_path = THRESHOLD_DIR / f"receipt_{index:02d}_adaptive.png"

    save_global = cv2.imwrite(str(global_output_path), global_thresh)
    save_adaptive = cv2.imwrite(str(adaptive_output_path), adaptive_thresh)

    if not save_global or not save_adaptive:
        raise IOError(f"Could not save thresholded images for receipt {index}")

    processed_count += 1

    # Step 7: Show comparison for each receipt
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(blurred_image, cmap="gray", vmin=0, vmax=255)
    axes[0].set_title(f"Receipt {index} - Blurred")
    axes[0].axis("off")

    axes[1].imshow(global_thresh, cmap="gray", vmin=0, vmax=255)
    axes[1].set_title(f"Receipt {index} - Global Threshold")
    axes[1].axis("off")

    axes[2].imshow(adaptive_thresh, cmap="gray", vmin=0, vmax=255)
    axes[2].set_title(f"Receipt {index} - Adaptive Threshold")
    axes[2].axis("off")

    plt.suptitle(
        f"Thresholding Comparison - Receipt {index}",
        fontsize=14,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()

    print(f"\nReceipt {index}")
    print("Input File          :", image_path.name)
    print("Global Output File  :", global_output_path.name)
    print("Adaptive Output File:", adaptive_output_path.name)

# Step 8: Final verification
print("\n" + "=" * 65)

print("Images Processed:", processed_count)
print("Global Images Saved  :", len(list(THRESHOLD_DIR.glob("*_global.png"))))
print("Adaptive Images Saved:", len(list(THRESHOLD_DIR.glob("*_adaptive.png"))))
print("Output Directory:", THRESHOLD_DIR)

assert processed_count == len(blur_files)

print("\nPHASE 2 - CELL 3 COMPLETED SUCCESSFULLY!")
```

```python

# ============================================================
# Crack404 AI/ML Internship - Week 3
# Project: Automated Invoice & Receipt Digitization
#
# PHASE 2 - CELL 4
# Select and Save Final Cleaned Images
# ============================================================

from pathlib import Path
import shutil
import cv2
import matplotlib.pyplot as plt

# Step 1: Define directories
PROJECT_DIR = Path(r"D:\Ai_Ml_intren\week_3")

THRESHOLD_DIR = PROJECT_DIR / "outputs" / "phase2_threshold"
CLEANED_DIR = PROJECT_DIR / "cleaned_images"

CLEANED_DIR.mkdir(parents=True, exist_ok=True)

# Step 2: Collect thresholded images
global_files = sorted(THRESHOLD_DIR.glob("*_global.png"))
adaptive_files = sorted(THRESHOLD_DIR.glob("*_adaptive.png"))

print("=" * 60)
print("PHASE 2 - CELL 4: FINAL CLEANED IMAGES")
print("=" * 60)

print("\nGlobal Images:", len(global_files))
print("Adaptive Images:", len(adaptive_files))

assert len(global_files) == len(adaptive_files)
assert 3 <= len(adaptive_files) <= 5

# Step 3: Choose preprocessing method
# Initial choice: adaptive for all receipts.
# Change individual values to "global" if needed.

method_selection = {
    1: "adaptive",
    2: "adaptive",
    3: "adaptive",
    4: "adaptive",
    5: "adaptive"
}

# Step 4: Save selected cleaned images
saved_files = []

for index in range(1, len(adaptive_files) + 1):

    method = method_selection[index]

    source_path = (
        THRESHOLD_DIR /
        f"receipt_{index:02d}_{method}.png"
    )

    output_path = (
        CLEANED_DIR /
        f"receipt_{index:02d}_cleaned.png"
    )

    if not source_path.exists():
        raise FileNotFoundError(source_path)

    shutil.copy2(source_path, output_path)

    saved_files.append(output_path)

    print(f"\nReceipt {index}")
    print("Selected Method:", method)
    print("Saved File:", output_path.name)

# Step 5: Visualize final cleaned images
fig, axes = plt.subplots(
    1,
    len(saved_files),
    figsize=(17, 5),
    squeeze=False
)

for index, image_path in enumerate(saved_files):

    image = cv2.imread(
        str(image_path),
        cv2.IMREAD_GRAYSCALE
    )

    if image is None:
        raise ValueError(
            f"Cannot read: {image_path.name}"
        )

    axes[0, index].imshow(
        image,
        cmap="gray",
        vmin=0,
        vmax=255
    )

    axes[0, index].set_title(
        f"Cleaned Receipt {index + 1}"
    )

    axes[0, index].axis("off")

plt.suptitle(
    "Week 3 - Final Cleaned Receipt Images",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# Step 6: Final verification
print("\n" + "=" * 60)
print("Images Saved:", len(saved_files))
print("Output Directory:", CLEANED_DIR)

assert all(file.exists() for file in saved_files)

print("\nPHASE 2 - CELL 4 COMPLETED SUCCESSFULLY!")
print("PHASE 2: IMAGE PREPROCESSING COMPLETED!")
```
