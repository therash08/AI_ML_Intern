```python

import json
from pathlib import Path

import joblib
import numpy as np

PROJECT_ROOT = Path(r"D:\Ai_Ml_intren")

WEEK2_DIR = PROJECT_ROOT / "week_2"
WEEK5_DIR = PROJECT_ROOT / "week_5"

MODEL_PATH = WEEK2_DIR / "model_v1.joblib"
REPORT_DIR = WEEK5_DIR / "outputs" / "reports"

REPORT_DIR.mkdir(parents=True, exist_ok=True)

MODEL_REPORT_PATH = (
    REPORT_DIR / "phase1_model_inspection.json"
)

if not MODEL_PATH.is_file():
    raise FileNotFoundError(
        f"Model file not found: {MODEL_PATH}"
    )

if MODEL_PATH.stat().st_size == 0:
    raise ValueError("Model file is empty.")

model_package = joblib.load(MODEL_PATH)

if not isinstance(model_package, dict):
    raise TypeError("Unexpected model package format.")

required_keys = {
    "model",
    "threshold",
    "feature_names"
}

missing_keys = required_keys.difference(
    model_package.keys()
)

if missing_keys:
    raise ValueError(
        f"Missing model package keys: {missing_keys}"
    )

model = model_package["model"]
threshold = float(model_package["threshold"])

feature_names = list(
    model_package["feature_names"]
)

if not callable(getattr(model, "predict_proba", None)):
    raise ValueError(
        "Model does not support predict_proba()."
    )

if not 0 <= threshold <= 1:
    raise ValueError("Invalid classification threshold.")

if len(feature_names) != 22:
    raise ValueError(
        f"Expected 22 features, found {len(feature_names)}."
    )

if len(set(feature_names)) != len(feature_names):
    raise ValueError("Duplicate feature names detected.")

if not all(isinstance(name, str) for name in feature_names):
    raise TypeError("Feature names must be strings.")

model_classes = getattr(model, "classes_", None)

if model_classes is not None:
    model_classes = np.asarray(model_classes).tolist()

model_input_features = getattr(
    model,
    "n_features_in_",
    None
)

if model_input_features is not None:
    model_input_features = int(model_input_features)

    if model_input_features != len(feature_names):
        raise ValueError(
            "Model input dimension does not match feature names."
        )

model_type = (
    f"{type(model).__module__}."
    f"{type(model).__name__}"
)

model_report = {
    "model_file": MODEL_PATH.name,
    "model_type": model_type,
    "model_size_kb": round(
        MODEL_PATH.stat().st_size / 1024,
        2
    ),
    "package_keys": list(model_package.keys()),
    "classification_threshold": threshold,
    "total_features": len(feature_names),
    "feature_names": feature_names,
    "model_input_features": model_input_features,
    "model_classes": model_classes,
    "predict_proba_available": True,
    "status": "VERIFIED"
}

with open(
    MODEL_REPORT_PATH,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        model_report,
        file,
        indent=4,
        ensure_ascii=False
    )

with open(
    MODEL_REPORT_PATH,
    "r",
    encoding="utf-8"
) as file:
    saved_report = json.load(file)

assert saved_report["status"] == "VERIFIED"
assert saved_report["total_features"] == 22
assert saved_report["predict_proba_available"] is True
assert MODEL_REPORT_PATH.is_file()

print("=" * 60)
print("PHASE 1 - CELL 3: MODEL LOADING & INSPECTION")
print("=" * 60)

print("\nMODEL INFORMATION")
print("Model File:", MODEL_PATH.name)
print("Model Type:", model_type)
print("Model Size:", model_report["model_size_kb"], "KB")
print("Model Loaded: PASS")

print("\nMODEL PACKAGE")
print("Package Keys:", list(model_package.keys()))
print("Classification Threshold:", threshold)
print("Total Input Features:", len(feature_names))
print("Prediction Probability Available:", True)

print("\nMODEL CLASSES")
print("Classes:", model_classes)

print("\nINPUT FEATURE NAMES")

for index, feature in enumerate(feature_names, start=1):
    print(f"{index:02d}. {feature}")

print("\nMODEL DIMENSION VERIFICATION")
print("Expected Features:", len(feature_names))
print("Model Input Features:", model_input_features)
print("Feature Verification: PASS")

print("\nMODEL REPORT")
print("Saved File:", MODEL_REPORT_PATH)
print("Report Exists:", MODEL_REPORT_PATH.is_file())

print("\nALL MODEL INSPECTION CHECKS PASSED!")
print("PHASE 1 - CELL 3 COMPLETED SUCCESSFULLY!")
```

```python

import json
from pathlib import Path

import joblib
import pandas as pd

PROJECT_ROOT = Path(r"D:\Ai_Ml_intren")
MODEL_PATH = PROJECT_ROOT / "week_2" / "model_v1.joblib"
REPORT_DIR = PROJECT_ROOT / "week_5" / "outputs" / "reports"
REPORT_DIR.mkdir(parents=True, exist_ok=True)

REPORT_PATH = REPORT_DIR / "phase1_prediction_verification.json"

model_package = joblib.load(MODEL_PATH)

model = model_package["model"]
threshold = float(model_package["threshold"])
feature_names = list(model_package["feature_names"])

new_customer = {
    "gender": "Male",
    "SeniorCitizen": 0,
    "Partner": "No",
    "Dependents": "No",
    "tenure": 4,
    "PhoneService": "Yes",
    "MultipleLines": "Yes",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 95.50,
    "TotalCharges": 382.00
}

customer_df = pd.DataFrame([new_customer])

customer_df["EstimatedLifetimeBilling"] = (
    customer_df["tenure"] * customer_df["MonthlyCharges"]
)

customer_df["TenureGroup"] = pd.cut(
    customer_df["tenure"],
    bins=[-1, 12, 24, 48, float("inf")],
    labels=[
        "New Customers (0-12)",
        "Short-Term (13-24)",
        "Mid-Term (25-48)",
        "Long-Term (49+)"
    ]
)

service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

customer_df["NumAddOnServices"] = (
    customer_df[service_columns].eq("Yes").sum(axis=1)
)

missing_features = [
    name for name in feature_names
    if name not in customer_df.columns
]

if missing_features:
    raise ValueError(
        f"Missing required features: {missing_features}"
    )

model_input_df = customer_df[feature_names].copy()

if model_input_df.shape != (1, 22):
    raise ValueError("Invalid model input shape.")

if model_input_df.isna().any().any():
    raise ValueError("Missing values detected in model input.")

model_classes = list(model.classes_)

if 1 not in model_classes:
    raise ValueError("Positive class 1 not found.")

positive_class_index = model_classes.index(1)

churn_probability = float(
    model.predict_proba(model_input_df)[0, positive_class_index]
)

prediction = int(churn_probability >= threshold)

prediction_label = (
    "Churn" if prediction == 1 else "No Churn"
)

expected_probability_percent = 83.01
actual_probability_percent = round(
    churn_probability * 100, 2
)

probability_match = (
    actual_probability_percent == expected_probability_percent
)

threshold_match = threshold == 0.36
prediction_match = prediction_label == "Churn"
feature_match = model_input_df.columns.tolist() == feature_names

verification_status = all([
    probability_match,
    threshold_match,
    prediction_match,
    feature_match
])

report = {
    "model_file": MODEL_PATH.name,
    "total_features": len(feature_names),
    "input_shape": list(model_input_df.shape),
    "churn_probability": churn_probability,
    "churn_probability_percent": actual_probability_percent,
    "expected_probability_percent": expected_probability_percent,
    "classification_threshold": threshold,
    "prediction": prediction,
    "prediction_label": prediction_label,
    "probability_match": probability_match,
    "threshold_match": threshold_match,
    "prediction_match": prediction_match,
    "feature_order_match": feature_match,
    "status": "PASS" if verification_status else "FAIL"
}

with open(REPORT_PATH, "w", encoding="utf-8") as file:
    json.dump(report, file, indent=4)

print("=" * 60)
print("PHASE 1 - CELL 5: PREDICTION REPRODUCTION")
print("=" * 60)

print("\nMODEL INFORMATION")
print("Model:", type(model).__name__)
print("Features:", len(feature_names))
print("Input Shape:", model_input_df.shape)

print("\nCUSTOMER INFORMATION")
print("Tenure:", new_customer["tenure"])
print("Contract:", new_customer["Contract"])
print("Monthly Charges:", new_customer["MonthlyCharges"])

print("\nPREDICTION RESULTS")
print("Churn Probability:", f"{churn_probability:.2%}")
print("Classification Threshold:", threshold)
print("Predicted Class:", prediction)
print("Final Prediction:", prediction_label)

print("\nWEEK 2 COMPARISON")
print("Expected Probability:", f"{expected_probability_percent:.2f}%")
print("Actual Probability:", f"{actual_probability_percent:.2f}%")
print("Probability Match:", "PASS" if probability_match else "FAIL")
print("Threshold Match:", "PASS" if threshold_match else "FAIL")
print("Prediction Match:", "PASS" if prediction_match else "FAIL")
print("Feature Order Match:", "PASS" if feature_match else "FAIL")

print("\nVERIFICATION REPORT")
print("Report Path:", REPORT_PATH)
print("Report Exists:", REPORT_PATH.is_file())

if not verification_status:
    raise AssertionError("Prediction reproduction verification failed.")

print("\nALL PREDICTION VERIFICATION CHECKS PASSED!")
print("PHASE 1 - CELL 5 COMPLETED SUCCESSFULLY!")
```
